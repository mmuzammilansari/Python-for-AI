### The four pillars, on one page

| Pillar | In one line | Key tools today |
|---|---|---|
| **Encapsulation** | Data + its guard together, control access | `_protected`, `__private`, `@property` |
| **Abstraction** | Simple interface, hidden complexity | `abc.ABC`, `@abstractmethod` |
| **Inheritance** | A class reuses another | `class Child(Parent):` → Class 11 |
| **Polymorphism** | Same call, many behaviours | overriding (yes), overloading (use `*args`) |
